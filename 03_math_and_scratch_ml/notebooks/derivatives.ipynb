```python
import numpy as np
X = np.random.randint(0, 10, (2, 2))
W = np.random.randint(0, 10, (2, 2))
b = np.random.randint(0, 10, (2,))
```

```python
y = X @ W + b
y_true = np.random.randint(0, 10, y.shape)
# mse_loss = ((y_true - y) ** 2) / len(y_true) # This is not what we will use (it's homework for you)
loss = y.sum()
dL_dy = np.ones_like(y)
```

```python
def compute_gradients(X, dL_dy):
    dL_dW = X.T @ dL_dy
    # dL_dX = dL_dy @ W.T
    dL_db = dL_dy.sum(axis = 1)
    return dL_dW

def perform_gradient_descent(W, b, dL_dW, dL_db, lr = 0.01):
    # --------------- Gradient Descent --------------- 
    lr = 0.01
    W = W - dL_dW * lr
    # b = b - dL_db * lr
    return W

def predict(X, W, b):
    return X @ W + b

for i in range(100):
    dL_dW, dL_db = compute_gradients(X, dL_dy)
    W, b = perform_gradient_descent(W, b, dL_dW, dL_db)

    if i % 30 == 0:
        y = predict(X, W, b)
        loss = y.sum()
        print(f"[epoch = {i} / {100}] | Loss = {loss:.3f}")
```

```python
import numpy as np
X = np.random.rand(2, 5)
W = np.random.rand(5, 1)
bias = np.random.rand(1,)
y_true = np.array([5, -1]).reshape(2, 1)
print(f"Original Predictions: {X @ W + bias}")
# print(f"Original Predictions: {X @ W}")
print("-" * 100)
print(f"True Targets: {y_true}")
print("-" * 100)
for i in range(10000):
    y_p = X @ W + bias # issue [node-2]                     <-------------- Forward pass
    # print(f"y_p.shape = {y_p.shape}")
    # y_p = X @ W
    loss_fn = (np.sum((y_true - y_p) ** 2)) / len(y_p)

    # -------------------------- Backward pass [Starts] ---------------------------
    dL_dy = (-2 / (len(y_p))) * (y_true - y_p) # Issue [node-1]
    # print(f"dL_dy.shape = {dL_dy.shape}")
    dL_dW = X.T @ dL_dy
    # print(f"dL_dW.shape = {dL_dW.shape}")
    dL_db = dL_dy.sum() # issue [node-3]
    # -------------------------- Backward pass [Ends] ------------------------------

    # print(f"dL_db.shape = {dL_db.shape}")
    # ------------------------- Optimizers starts -------------------------------
    W = W - 0.01 * dL_dW
    bias = bias - 0.01 * dL_db # issue [node-4]
    # ------------------------- Optimizers ends -------------------------------
    # print(f"updated_bias.shape = {bias.shape}")

    if i % 1000 == 0:
        print(f"Epochs [{i}/{10000}] | Loss = {loss_fn}")

print("-" * 100)
print(f"Final Predictions: {X @ W + bias}")
```
