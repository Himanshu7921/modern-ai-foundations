# Assignment: Implement Gradient Descent using MSE and MAE Loss Functions

Consider the linear model:

$$
Y = XW + b
$$

where:

* $X$ = Input Matrix
* $W$ = Weight Matrix (Learnable Parameter)
* $b$ = Bias Vector (Learnable Parameter)
* $Y$ = Model Output

---

## Part A: Mean Squared Error (MSE)

### Tasks

1. Generate random input data $X$, learnable parameters $W$ and $b$, and any ground truth values $Y_{true}$.

2. Implement the forward pass:

$$
Y_{pred} = XW + b
$$

3. Implement the Mean Squared Error (MSE) loss function:

$$
L_{MSE}
= 

\frac{1}{N}
\sum_{i=1}^{N}
(Y_{true}^{(i)} - Y_{pred}^{(i)})^2
$$

4. Derive and implement the gradients manually.

5. Implement gradient descent to update $W$ and $b$.

6. Train the model for multiple epochs and verify that the MSE loss decreases over time.

---

## Part B: Mean Absolute Error (MAE)

### Tasks

1. Using the same model and dataset, replace the loss function with Mean Absolute Error (MAE):

$$
L_{MAE}
=

\frac{1}{N}
\sum_{i=1}^{N}
\left|
Y_{true}^{(i)} - Y_{pred}^{(i)}
\right|
$$

2. Derive and implement the gradients manually.

3. Implement gradient descent to update $W$ and $b$.

4. Train the model for multiple epochs and verify that the MAE loss decreases over time.

---

## Requirements

* Use **NumPy only**.
* Do **not** use PyTorch, TensorFlow, JAX, Scikit-Learn, or any automatic differentiation library.
* Print the loss value periodically during training.
* Demonstrate that both MSE and MAE losses decrease during optimization.
* You may use any randomly generated ground truth values.
