## Variables and Datatypes

```python
# In Java/C++, variables must be explicitly defined (e.g., int age = 19;)
# Python uses implicit definition—it detects the type automatically

age = 19
print(type(age)) # Output: <class 'int'>

lr = 0.01
print(type(lr)) # Output: <class 'float'>

name = "my_name"
print(type(name)) # Output: <class 'str'>

is_even = False
print(type(is_even)) # Output: <class 'bool'>

# Note: complex type exists but is mostly used in quantum models/physics
```

## Collections in Python: Lists

```python
# Unlike Java arrays which strictly enforce a single data type, 
# Python lists are highly flexible and can store mixed data types.

my_list_1 = ["my_name_1", "my_name_2", "my_name_3", 1.3, 0, 7, True, False]
print("Complete list:", my_list_1)

# Positive Indexing (Starts from 0)
print("First element:", my_list_1[0])
print("Third element:", my_list_1[2])

# Negative Indexing (Starts from -1 from the back)
print("Last element:", my_list_1[-1])
print("Second last element:", my_list_1[-2])
```

## Slicing and Reversing Lists

```python
# Syntax: list[start : end : step] 
# Remember: start index is INCLUDED, end index is EXCLUDED

# Basic slicing
print("Elements from index 0 to 2:", my_list_1[0:3])

# Slicing with a step skipping entries
print("Every 3rd element up to last:", my_list_1[0:-1:3])

# Method 1: Reversing a list completely
print("Fully reversed list:", my_list_1[::-1])

# Method 2: Reverse excluding the first element (index 0)
print("Reversed till index 1:", my_list_1[-1:0:-1])
```

## Mutability and Immutability Test

```python
# Testing List Mutability
ages = [18, 21]
ages[0] += 1
ages[1] += 1
print("Updated list entries:", ages) # Lists are mutable

# Testing Dictionary Mutability
my_stored_information = {
    "Rohan": {"age": 19, "school": "XYZ", "school_id": "Y"},
    "khrisha": {"age": 18, "school": "ABC", "school_id": "123@456@XYZ"}
}
my_stored_information["khrisha"]["age"] += 1
print("Updated age via key lookup:", my_stored_information["khrisha"]["age"]) # Dictionaries are mutable

# Testing Tuple Immutability
try:
    my_tuple = ((1, 2), (4, 5))
    my_tuple[0] = (9, 10) # This will crash the execution
except TypeError as e:
    print("Tuples cannot be altered:", e)
```

## Sets and String Formatting

```python
# String formatting using f-strings
name = "Khrisha"
age = 18
print(f"Hello my name is {name} and my age is {age}")

# Sets remove duplicate records and do not maintain sequence order
users_list = ["Rohan", "Mohan", "Tanish", "Mohan"]
users_set = set(users_list)
print("Unique users set:", users_set)

# Sets are mutable
users_set.add("ABC")
users_set.update(["3", "4"])
print("Updated set items:", users_set)
```

## Loops and If-Else Statements

```python
# If-else control flow block
voting_age_limit = 18
age = 17 + 1

if age >= voting_age_limit:
    print("Eligible!")
else:
    print("Not Eligible!")

# Iterating through arrays/lists using native loops
my_list = ["ABC", "DEF", 123, 456]

print("--- For loop without indexing ---")
for item in my_list:
    print(item)

print("--- While loop iteration ---")
idx = 0
while idx != len(my_list):
    print(my_list[idx])
    idx += 1
```

## Functions

```python
# Python uses the 'def' keyword to define modular function logic

# 1. Function taking no arguments
def say_hello():
    print("Hello")

say_hello()

# 2. Function taking arguments but returning nothing
def say_hello_to(name):
    print(f"Hello, {name}")

say_hello_to("Khrisha")

# 3. Function taking arguments and returning a value
def get_name_length(name):
    print(f"Processing name: {name}")
    return len(name)

count = get_name_length("ABC")
print("Calculated word length:", count)
```

## Object-Oriented Programming (OOP) Foundations

Object-Oriented Programming organizes code into reusable blocks using **Classes** and **Objects**:
* **Class:** A blueprint or template that defines attributes (data) and methods (actions).
* **Object:** A physical instance of a class containing real values.
* **`__init__` (Constructor):** A special initializer method that sets up initial attributes when an object is instantiated.
* **`self`:** Refers to the current instance of the class being processed (similar to `this` in Java/C++).

```python
# Defining simple blueprints (Classes)
class Dog:
    # The constructor method initializes the object's properties
    def __init__(self, name):
        self.name = name # Storing the name in the instance variable
        
    def bark(self):
        return f"{self.name} barking.."

class Cat:
    def sound(self, name):
        return f"{name} meow..."

# Creating instances (Objects) of our classes
dog = Dog("Tommy")
print(dog.bark()) # Output: Tommy barking..

print(f"Dog name = {dog.name}") # Accessing attributes directly
```

## Inheritance

**Inheritance** allows a child class to inherit attributes and methods from a parent clas. This prevents code repetition and establishes a clear structural hierarchy.
* **`super()`:** A built-in function used to call methods (most commonly the constructor) from the parent class inside the child class.
* **Types of Inheritance:** Single, Multilevel, Multiple, and Hierarchical.

```python
# Parent Class
class Car:
    def __init__(self, car_name, engine_type, wheel_control):
        self.car_name = car_name
        self.engine_type = engine_type
        self.wheel_control = wheel_control

    def drive(self):
        print(f"Driving car = {self.car_name}")

    def print_details(self):
        print(f"Car Name = {self.car_name}")
        print(f"Engine Type = {self.engine_type}")
        print(f"Wheel control = {self.wheel_control}")

# Child Class inheriting from Car (Single Inheritance)
class BMWCar(Car):
    def __init__(self, car_name, engine_type, wheel_control):
        # super() invokes the constructor of the parent 'Car' class
        super().__init__(car_name, engine_type, wheel_control)
        super().print_details() # Automatically execute print details on creation

    def change_engine(self, new_engine_type):
        self.engine_type = new_engine_type

# Child Class using 'pass' inherits everything without modification
class AudiCar(Car):
    pass

# Initializing the child object
car = BMWCar(car_name="BMW-M4", engine_type="v-8", wheel_control=2)

# Modifying inherited attribute state using a child method
car.change_engine("v-10")
print("Updated Engine Type:", car.engine_type) # Output: 'v-10'
```

## Polymorphism

**Polymorphism** means "many forms".It allows different classes to share the same method names but exhibit completely unique operational behaviors. 
* **Method Overriding (Runtime Polymorphism):** When a child class redefines an inherited method from its parent class to change its behavior.

```python
# Base class representing a standard device connection
class Phone:
    def __init__(self, phone_name):
        self.name = phone_name

    def call(self):
        print("calling call() method of Phone class")
        print(f"Calling from {self.name}...")

# Subclass overriding the call method
class SmartPhone(Phone):
    def __init__(self, phone_name):
        super().__init__(phone_name)

    def video_call(self):
        print(f"Outgoing video call from {self.name}...")

    # Overriding the parent 'call' method entirely
    def call(self):
        print(f"Calling from {self.name}...")

# Baseline test using the parent class behavior
phone = Phone(phone_name="Nokia")
phone.call()

# Polymorphic behavior test using runtime method overriding
iphone = SmartPhone(phone_name="iphone")
iphone.video_call()
iphone.call() # Executes overridden version instead of Parent class method
```
