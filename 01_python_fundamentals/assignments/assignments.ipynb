```python
# ==============================================================================================
# QUESTION 1: The Secure Banking System (All 4 Pillars)
# ----------------------------------------------------------------------------------------------
# Task: Create a banking blueprint simulating multiple account types with strict data security.
#
# Requirements:
# 1. Abstraction: Create an abstract base class `BankAccount` (using the `abc` library) that 
#    enforces child classes to implement a `calculate_interest()` method.
# 2. Encapsulation: The `BankAccount` constructor must accept `account_holder` and `_balance`.
#    Keep the balance protected/private (using an underscore `_balance`) so it cannot be 
#    modified directly outside the class. Provide a `get_balance()` method to read it safely.
# 3. Inheritance: Create two child classes inheriting from `BankAccount`:
#    - `SavingsAccount`: Adds an `interest_rate` attribute (e.g., 0.04 for 4%).
#    - `CurrentAccount`: Adds an `overdraft_limit` attribute (e.g., 500).
# 4. Polymorphism (Method Overriding): Override the `calculate_interest()` method in both:
#    - In `SavingsAccount`, it computes: balance * interest_rate.
#    - In `CurrentAccount`, it returns 0 (current accounts do not earn interest).
# 5. Instantiation: Create one object for each account type, call `calculate_interest()`, 
#    and print out the account balances securely using your getter method.
# ==============================================================================================

# Write your solution in a new cell below this one
```

```python
from abc import ABC, abstractmethod
class BankAccount(ABC):
  def __init__(self, account_holder, _balance):
    self.accountholder=account_holder
    self._balance= _balance

  def get_balance(self):
    print(f"Name: {self.account_holder}")
    print(f"Current Balance: {self._balance}")

  @abstractmethod
  def calculate_interest(self):
    pass

 # __________________

class SavingsAccount(BankAccount):
  def __init__(self,account_holder, _balance, interest_rate, time):
    super().__init__(account_holder, _balance)
    self.account_holder= account_holder
    self._balance= _balance
    self.interest_rate= interest_rate
    self.time= time
  def calculate_interest(self):
    print(f"Interest gained is: {self.interest_rate * self._balance * self.time}")

 # __________________

class CurrentAccount(BankAccount):
  def __init__(self,account_holder, _balance, overdraft_limit):
    super().__init__(account_holder, _balance)
    self.account_holder= account_holder
    self._balance= _balance
    self.overdraft_limit= overdraft_limit
  def calculate_interest(self):
    return 0

 # __________________

sv= SavingsAccount("Khrisha",200000,0.04, 2)
sv.get_balance()
sv.calculate_interest()

ca= CurrentAccount("Khrisha", 200000, 100000)
ca.get_balance()
print(f"Interest gained: {ca.calculate_interest()}")
```

```python
# ==============================================================================================
# QUESTION 2: Machine Learning Layer Architecture Pipeline
# ----------------------------------------------------------------------------------------------
# Task: Design an abstract model layer system imitating how deep learning frameworks 
#       (like PyTorch) handle forward passes and parameter counts across different layer types.
#
# Requirements:
# 1. Abstraction: Create an abstract base class `Layer` with an abstract method `forward(x)`.
# 2. Encapsulation: The `Layer` class must store a protected string `layer_name` and a private 
#    integer `__parameters` (simulating the weight count). Create a getter method to safely 
#    access the private parameter count.
# 3. Inheritance & Polymorphism: Create two child classes that inherit from `Layer`:
#    - `LinearLayer`: Its `forward(x)` method should print out a simulated dot product text like 
#      "Computing matrix multiplication: X * W + b".
#    - `ReLULayer`: Its `forward(x)` method should print out a simulated activation function text 
#      like "Applying non-linear transformation: max(0, X)".
# 4. Execution: Create a list containing one `LinearLayer` instance (with 5000 parameters) 
#    and one `ReLULayer` instance (with 0 parameters). Loop through this list to execute the 
#    `forward(x)` method on each layer object and print their parameter counts.
# ==============================================================================================

# Write your solution in a new cell below this one
```

```python
from abc import ABC, abstractmethod
# 1. Abstraction [Done]
# 2. Encapsulation [Done]
# 3. Inheritance & Polymorphism [Done]
# 4. Execution [Done]

class Layer(ABC):
    def __init__(self, layer_name, parameters):
        self._layer_name = layer_name # Protected Variable
        self.__parameters = parameters # Private Variable

    @abstractmethod
    def forward(self, x):
        pass

    # Getter Method, for safely accessing this private variable "__parameters"
    def get_parameters(self):
        return self.__parameters

class LinearLayer(Layer):
    def __int__(self, layer_name, parameters):
        super().__init__(layer_name, parameters)
        self._layer_name = layer_name
    
    def forward(self, x):
        return "Computing matrix multiplication: X @ W + b"

class ReLULayer(Layer):
    def __init__(self, layer_name, parameters):
        super().__init__(layer_name, parameters)
        self._layer_name = layer_name
    
    def forward(self, x):
        return "Applying non-linear transformation: max(0, X)"
    
linear_layer = LinearLayer(layer_name = "Linear Layer", parameters = 5000)
relu_layer = ReLULayer(layer_name = "ReLU Layer", parameters = 0)

nn_layers = [linear_layer, relu_layer]
for layer in nn_layers:
    print(f"{layer.forward('x')} | Parameter Count: {layer.get_parameters()}")
```
